```python
import numpy as np
import math
import matplotlib.pyplot as plt
from sklearn import linear_model

%matplotlib inline
def gradient_descent(x, y):
    m_curr = b_curr = 0
    iterations = 1000
    n = len(x)
    learning_rate = 0.02
    cost = 0

    plt.scatter(x, y, color='red', marker='+')
    for i in range(iterations):
        # y = mx + b
        y_predicted = m_curr * x + b_curr
        plt.plot(x, y_predicted, color='green')
        prev_cost = cost
        cost = (1 / n) * sum([val**2 for val in (y - y_predicted)])

        if math.isclose(prev_cost, cost, rel_tol=1e-20):
            break

        m_derivative = -(2 / n) * sum(x * (y - y_predicted))
        b_derivative = -(2 / n) * sum(y - y_predicted)

        m_curr = m_curr - learning_rate * m_derivative
        b_curr = b_curr - learning_rate * b_derivative

        # print(
        #     f"m: {round(m_curr, 4)}, b: {round(b_curr, 4)}, prev_cost: {round(prev_cost, 4)}, cost: {round(cost, 4)}, is_close {math.isclose(prev_cost, cost)}, i: {i}"
        # )

    # print('gradient_descent',m_curr, b_curr)
    return m_curr, b_curr
```

```python
import pandas as pd

def predict_using_sklearn():
    df = pd.read_csv('test_scores.csv')

    regression_model = linear_model.LinearRegression()
    regression_model.fit(df[['math']], df.cs)
        
    print('regression_model',regression_model.coef_, regression_model.intercept_)
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 7, 9, 11, 13])

gradient_descent(x, y)
```
